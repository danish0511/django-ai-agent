```python
from ai.agents import get_document_agent
```

```python
agent = get_document_agent()
```

```python
import uuid

config = {"configurable": {"user_id": "1", "thread_id": uuid.uuid4()}}

response = agent.invoke(
    {"messages": 
        [
            {
                "role": "user",
                "content": "what are my recent documents?"
            }
        ]
    },
    config
)
```

```python
for msg in response["messages"]:
    print(msg.content)
```
